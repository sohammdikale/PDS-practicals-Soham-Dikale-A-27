## **1. Difference between Binomial, Poisson, and Normal Distributions**

Probability distributions describe how values of a random variable are distributed. The **Binomial, Poisson, and Normal distributions** are among the most important in statistics and are used to model different types of phenomena.

### **Detailed Comparison Table**

To understand their distinctions, let's look at a detailed comparison:

| Feature            | Binomial Distribution                                       | Poisson Distribution                                  | Normal Distribution                                 |
| :----------------- | :---------------------------------------------------------- | :---------------------------------------------------- | :-------------------------------------------------- |
| **Type**           | Discrete                                                    | Discrete                                              | Continuous                                          |
| **Purpose**        | Models the number of successes in a fixed number of independent Bernoulli (success/failure) trials. | Models the number of times an event occurs in a fixed interval of time or space, especially for rare events. | Models continuous data, often found in nature and social sciences, characterized by its symmetric, bell-shaped curve. |
| **Parameters**     | `n` (number of trials), `p` (probability of success on each trial) | `λ` (lambda: the average rate of events in the given interval) | `μ` (mu: mean, center of the distribution), `σ` (sigma: standard deviation, spread of the distribution) |
| **Possible Values** | Integers from 0 to `n` (inclusive)                          | Non-negative integers (0, 1, 2, ...) extending to infinity | Any real number from negative infinity to positive infinity |
| **Shape**          | Can be symmetric (if p=0.5), positively skewed (if p<0.5), or negatively skewed (if p>0.5) | Positively skewed for small λ, approaches symmetry as λ increases | Perfectly symmetric, bell-shaped                      |
| **Key Assumption** | Fixed number of trials, independent trials, only two outcomes per trial (success/failure), constant probability of success | Events occur independently, average rate of occurrence is constant over the interval, events are rare | Data is continuous, clustered around the mean, symmetrical spread |
| **Real-World Example** | The number of defective items in a batch of 100 produced items, given the probability of one item being defective. | The number of customers arriving at a checkout counter in one hour. | The heights of adult males in a specific population. |

### **Formulas**

Understanding the mathematical formulas helps in computing probabilities and grasping the underlying mechanics of each distribution.

#### **Binomial Distribution (Probability Mass Function)**

This formula calculates the probability of getting exactly `k` successes in `n` trials.

$$P(X=k)=\binom{n}{k} p^k (1-p)^{n-k}$$

Where:
*   $P(X=k)$ is the probability of exactly $k$ successes.
*   $\binom{n}{k} = \frac{n!}{k!(n-k)!}$ is the binomial coefficient, representing the number of ways to choose $k$ successes from $n$ trials.
*   $n$ is the total number of trials.
*   $k$ is the number of successful outcomes.
*   $p$ is the probability of success on a single trial.
*   $(1-p)$ is the probability of failure on a single trial (often denoted as $q$).

#### **Poisson Distribution (Probability Mass Function)**

This formula calculates the probability of exactly `k` events occurring in a fixed interval of time or space, given the average rate $\lambda$.

$$P(X=k)=\frac{\lambda^k e^{-\lambda}}{k!}$$

Where:
*   $P(X=k)$ is the probability of exactly $k$ events occurring.
*   $\lambda$ (lambda) is the average rate of event occurrences in the given interval.
*   $k$ is the actual number of events that occur.
*   $e$ is Euler's number (approximately 2.71828).
*   $k!$ is the factorial of $k$.

#### **Normal Distribution (Probability Density Function)**

This formula describes the probability distribution for a continuous random variable. The area under the curve gives the probability.

$$f(x)=\frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

Where:
*   $f(x)$ is the probability density function for the normal distribution.
*   $\mu$ (mu) is the mean of the distribution.
*   $\sigma$ (sigma) is the standard deviation of the distribution.
*   $\pi$ (pi) is approximately 3.14159.
*   $e$ is Euler's number (approximately 2.71828).
*   $x$ is the value of the random variable.

## **2. Real-World Significance of the Poisson Distribution**

The **Poisson distribution** holds immense importance in various real-world scenarios, particularly when dealing with **rare events that occur over a fixed interval of time or space**. Its significance stems from its ability to model and predict the occurrence of such events, which is crucial for planning, risk assessment, and resource management.

### **Key Idea and Characteristics**

The Poisson distribution is a discrete probability distribution that expresses the probability of a given number of events occurring in a fixed interval of time or space if these events occur with a known constant mean rate and independently of the time since the last event.

Its key assumptions are:

*   **Independence:** Events occur independently of each other.
*   **Constant Rate:** The average rate of event occurrence ($\lambda$) is constant over the interval.
*   **Rarity:** Events are rare within the given interval.

### **Real-World Applications and Practical Examples**

The applications of the Poisson distribution are widespread across numerous fields:

| Field                     | Example Scenario                                        | Practical Implication                                        |
| :------------------------ | :------------------------------------------------------ | :----------------------------------------------------------- |
| **Telecommunications**    | Number of phone calls received by a call center per minute/hour. | Staffing optimization for call centers to minimize waiting times. |
| **Healthcare**            | Number of emergency room admissions in a hospital per day/shift. | Resource allocation (doctors, nurses, beds) to handle patient influx. |
| **Traffic Engineering**   | Number of accidents at a specific intersection per week/month. | Designing safer roads or implementing traffic control measures. |
| **Banking/Finance**       | Number of ATM withdrawals or online transactions per hour. | Managing server capacity and detecting unusual transaction patterns (fraud). |
| **E-commerce/Web Analytics** | Number of clicks on an advertisement or website visits per second/minute. | Optimizing ad placement, server load balancing, and website performance. |
| **Manufacturing/Quality Control** | Number of defects per unit area of a fabric or per batch of products. | Implementing quality checks and identifying production issues. |
| **Biology/Ecology**       | Number of mutations in a DNA strand per unit length, or the number of rare species observed in a defined area. | Understanding genetic variations or population dynamics. |
| **Insurance**             | Number of insurance claims filed per month for a specific type of event. | Calculating premiums and managing risk portfolios. |

### **Why It Matters (Impact and Importance)**

*   **Resource Planning:** Businesses can use Poisson distribution to forecast demand (e.g., number of customers, service requests) and allocate resources (staff, inventory, server capacity) more effectively, reducing costs and improving service quality.
*   **Risk Management:** It helps in quantifying the probability of rare but critical events (e.g., system failures, accidents), allowing organizations to develop contingency plans and mitigate risks.
*   **Queueing Theory:** A fundamental tool in queueing models, helping to design efficient systems where customers or items wait for service.
*   **Quality Control:** Manufacturers use it to monitor the number of defects, ensuring product quality and identifying production process deviations.
*   **Forecasting:** Although simple, it provides a baseline for forecasting event occurrences in situations where average rates are stable.

In essence, the Poisson distribution provides a robust framework for understanding and making predictions about random, rare events, which is invaluable for operational efficiency and strategic decision-making.

## **3. Central Limit Theorem (CLT) and Relation to Normal Distribution**

The **Central Limit Theorem (CLT)** is one of the most fundamental and powerful concepts in statistics. It provides a bridge between sample statistics and population parameters, particularly concerning the normal distribution.

### **Definition of the Central Limit Theorem**

The Central Limit Theorem states that:

> Regardless of the shape of the original population distribution (whether it's normal, uniform, skewed, etc.), if you take sufficiently large random samples from that population, the sampling distribution of the sample means will be approximately normally distributed. Furthermore, the mean of this sampling distribution will be equal to the population mean ($\mu$), and its standard deviation (known as the standard error) will be equal to the population standard deviation ($\sigma$) divided by the square root of the sample size ($\sqrt{n}$).

### **Mathematical Representation**

The theorem implies that if we have a population with mean $\mu$ and standard deviation $\sigma$, and we take sufficiently large samples of size $n$, the sample mean $\bar{X}$ will have a distribution that can be approximated by a normal distribution with:

*   Mean: $E(\bar{X}) = \mu$
*   Standard Deviation (Standard Error): $SE = \frac{\sigma}{\sqrt{n}}$

For a large sample size ($n$), the standardized sample mean (Z-score) will follow a standard normal distribution:

$$Z=\frac{\bar{X}-\mu}{\sigma/\sqrt{n}}$$

Where:
*   $Z$ is the standard normal variable.
*   $\bar{X}$ is the sample mean.
*   $\mu$ is the population mean.
*   $\sigma$ is the population standard deviation.
*   $n$ is the sample size.

### **Key Points and Conditions**

*   **Universality:** The CLT applies to almost any type of population distribution, which is its most remarkable feature.
*   **Sample Size:** The 'sufficiently large' sample size is crucial. Generally, a sample size of $n \ge 30$ is considered large enough for the CLT to hold true for most distributions. For distributions that are already somewhat symmetric, even smaller sample sizes might suffice.
*   **Random Sampling:** The samples must be drawn randomly and independently from the population.
*   **Resulting Distribution:** The sampling distribution of the sample means will be approximately normal.

### **Illustrative Example**

Consider a population of students' test scores that is highly skewed (e.g., many low scores, a few very high scores). This population distribution is non-normal.

1.  **Take a large number of random samples** (e.g., 50 samples of size 30 each) from this population.
2.  **Calculate the mean** for each of these 50 samples.
3.  **Plot a histogram of these 50 sample means.**

According to the CLT, this histogram of sample means will look approximately like a normal (bell-shaped) distribution, even though the original population of individual test scores was skewed. The mean of this distribution of sample means will be very close to the true mean of the entire student population.

### **Why the CLT is Important and its Relation to Normal Distribution**

The CLT is foundational for much of inferential statistics and demonstrates why the Normal Distribution is so prevalent:

| Benefit/Relation            | Explanation                                                                         |
| :-------------------------- | :---------------------------------------------------------------------------------- |
| **Simplifies Statistical Analysis** | It allows statisticians to use normal distribution-based statistical tests and confidence intervals even when the original population distribution is unknown or non-normal. This converts complex distributions into a more manageable, normally distributed form. |
| **Foundation for Hypothesis Testing** | Many common hypothesis tests (e.g., z-tests, t-tests) rely on the assumption of normally distributed sample means, which is justified by the CLT. |
| **Basis for Confidence Intervals** | It enables the construction of confidence intervals for population means, allowing us to estimate population parameters from sample data with a certain level of confidence. |
| **Understanding Data Variability** | The concept of standard error ($\sigma/\sqrt{n}$) derived from CLT helps quantify the variability of sample means, indicating how much sample means are expected to vary from the population mean. |
| **Widely Applicable**       | Its applicability to diverse initial distributions makes it a cornerstone for statistical inference in various scientific, economic, and social fields. |
| **Explains Natural Phenomena** | It helps explain why many measurements in the natural and social sciences (e.g., heights, IQ scores, measurement errors) tend to be normally distributed, as these are often the result of many small, independent factors adding up. |

In summary, the CLT is critical because it justifies the use of normal distribution-based methods in situations where the underlying population distribution is not normal, making statistical inference widely applicable and robust.

## **4. Why Normal Distribution is Widely Used in Machine Learning and Statistics**

The **Normal distribution**, also known as the Gaussian distribution, is arguably the most important probability distribution in statistics and machine learning. Its widespread use stems from its unique mathematical properties, its natural occurrence in many real-world phenomena, and the implications of the Central Limit Theorem.

### **Characteristics of the Normal Distribution**

To understand its popularity, let's recall its key features:

*   **Symmetry:** It is perfectly symmetrical around its mean. This means the left side of the curve is a mirror image of the right side.
*   **Bell-shaped Curve:** When plotted, it forms a characteristic bell shape.
*   **Measures of Central Tendency:** For a normal distribution, the Mean, Median, and Mode are all equal and located at the center of the distribution.
*   **Asymptotic Tails:** The tails of the curve extend infinitely in both directions, approaching, but never quite touching, the horizontal axis.
*   **Defined by Two Parameters:** It is completely defined by its mean ($\mu$) and standard deviation ($\sigma$). The mean shifts the curve horizontally, and the standard deviation determines its spread (width).

### **The 68–95–99.7 Rule (Empirical Rule)**

A particularly useful property of the normal distribution is the empirical rule, which describes the percentage of data that falls within a certain number of standard deviations from the mean:

*   Approximately **68%** of the data falls within **1 standard deviation** ($\pm 1\sigma$) of the mean.
*   Approximately **95%** of the data falls within **2 standard deviations** ($\pm 2\sigma$) of the mean.
*   Approximately **99.7%** of the data falls within **3 standard deviations** ($\pm 3\sigma$) of the mean.

This rule provides a quick way to understand the spread and typical range of data in a normally distributed dataset.

### **Reasons for Popularity in ML and Statistics**

The Normal distribution is widely used due to several compelling reasons:

| Reason                               | Explanation                                                                                                                                                                                                                                                              |
| :----------------------------------- | :--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Natural Occurrence**               | Many phenomena in nature and social sciences naturally tend to follow a normal distribution. Examples include human characteristics (height, weight, IQ scores), measurement errors, blood pressure, and many economic indicators.                                            |
| **Central Limit Theorem (CLT) Support** | As discussed earlier, the CLT dictates that the sampling distribution of sample means approaches a normal distribution as the sample size increases, regardless of the original population distribution. This makes the normal distribution a fundamental tool for inferential statistics. |
| **Mathematical Tractability and Simplicity** | The normal distribution has well-defined mathematical properties that make it easy to work with. It has a continuous probability density function that is differentiable and easy to manipulate, simplifying calculations of probabilities and statistical inference.                                |
| **Foundation for Statistical Inference** | Many classical statistical tests (e.g., t-tests, ANOVA, linear regression) assume that the data (or residuals) are normally distributed. This assumption simplifies their underlying mathematics and allows for robust hypothesis testing and confidence interval estimation. |
| **Machine Learning Algorithms Assumptions** | A significant number of machine learning algorithms either explicitly or implicitly assume that the underlying data follows a normal distribution. <br><br> *   **Linear Models (e.g., Linear Regression):** Often assume that the residuals (errors) are normally distributed. <br> *   **Gaussian Naive Bayes:** Explicitly assumes features are normally distributed (Gaussian). <br> *   **Linear Discriminant Analysis (LDA):** Assumes features follow a multivariate normal distribution. <br> *   **Principal Component Analysis (PCA):** While not strictly requiring normality, PCA often performs better when data is approximately normal or symmetrically distributed. <br> *   **Clustering (e.g., Gaussian Mixture Models):** Based on the idea of modeling data points as belonging to different Gaussian distributions. <br> *   **Outlier Detection:** The 68-95-99.7 rule is often used to identify outliers; data points far from the mean (e.g., beyond 2 or 3 standard deviations) are considered unusual. |
| **Parameter Estimation**             | For normally distributed data, the sample mean and standard deviation are excellent estimators of the population mean and standard deviation, respectively.                                                                                                                   |
| **Risk and Quality Control**         | In finance, stock returns are often approximated by a normal distribution for risk assessment. In manufacturing, it's used to model process variations and set quality control limits.                                                                                      |

### **Practical Example**

Consider the **IQ scores** of a large population. These scores are typically designed to follow a normal distribution with a mean ($\mu$) of 100 and a standard deviation ($\sigma$) of 15.

*   This means most people have an IQ around 100.
*   Using the 68-95-99.7 rule, we can say that about 68% of people have an IQ between 85 (100-15) and 115 (100+15).
*   About 95% have an IQ between 70 (100-2*15) and 130 (100+2*15).

This predictability and statistical power make the normal distribution an indispensable tool in both theoretical statistics and practical applications across various data-driven fields.

## **5. Skewness in Distributions**

While the Normal distribution is symmetric, many real-world datasets do not follow this perfect symmetry. **Skewness** is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. In simpler terms, it tells us about the shape of the distribution and whether it is stretched more to one side than the other.

### **Understanding Asymmetry**

*   A **symmetrical distribution** (like the Normal distribution) has its values distributed evenly around the mean. The left side is a mirror image of the right side.
*   A **skewed distribution**, on the other hand, has a 'tail' that is longer on one side, indicating that the data is concentrated more on one side and has fewer, but potentially larger or smaller, values extending towards the other side.

### **Types of Skewness**

There are three main types of skewness:

1.  **Positive Skew (Right-Skewed Distribution):**
    *   **Shape:** The tail of the distribution is longer on the right side. This means there are a few unusually large values (outliers) that pull the mean to the right.
    *   **Relationship of Measures of Central Tendency:** For a positively skewed distribution, typically **Mean > Median > Mode**.
    *   **Visual Representation:** Looks like the bulk of the data is on the left, with a long tail extending to the right.

2.  **Negative Skew (Left-Skewed Distribution):**
    *   **Shape:** The tail of the distribution is longer on the left side. This indicates the presence of a few unusually small values that pull the mean to the left.
    *   **Relationship of Measures of Central Tendency:** For a negatively skewed distribution, typically **Mean < Median < Mode**.
    *   **Visual Representation:** Looks like the bulk of the data is on the right, with a long tail extending to the left.

3.  **Zero Skew (Symmetrical Distribution):**
    *   **Shape:** The distribution is perfectly symmetrical around its mean. The left and right sides are exact mirror images.
    *   **Relationship of Measures of Central Tendency:** For a perfectly symmetrical distribution, **Mean = Median = Mode**.
    *   **Visual Representation:** A balanced shape, such as the bell curve of the Normal distribution.

### **Mathematical Measure of Skewness**

Skewness can be quantified using various coefficients. Pearson's first and second coefficients of skewness are simple to calculate, but the most common method involves the third standardized moment:

**Coefficient of Skewness (Fisher-Pearson coefficient of skewness):**

$$\gamma_1 = E\left[\left(\frac{X-\mu}{\sigma}\right)^3\right] = \frac{\mu_3}{\sigma^3}$$

Where:
*   $\gamma_1$ is the skewness coefficient.
*   $E[...]$ is the expected value.
*   $\mu$ is the mean of the distribution.
*   $\sigma$ is the standard deviation of the distribution.
*   $\mu_3$ is the third central moment (a measure of the shape of the distribution).

*   If $\gamma_1 > 0$, the distribution is positively skewed.
*   If $\gamma_1 < 0$, the distribution is negatively skewed.
*   If $\gamma_1 = 0$, the distribution is perfectly symmetrical.

### **Which Distributions Exhibit Skewness?**

| Distribution            | Skewness Characteristics                                                                                                                                                                                                                                                                                                             |
| :---------------------- | :----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Normal Distribution** | Exhibits **zero skewness** by definition. It is a perfectly symmetrical, bell-shaped distribution where the mean, median, and mode are all equal.                                                                                                                                                                                         |
| **Poisson Distribution** | Is inherently **positively skewed**, especially for small values of $\lambda$ (the average rate). As $\lambda$ increases, the Poisson distribution approaches symmetry and starts to resemble a Normal distribution. This positive skew reflects the rarity of events; it's more likely to observe a small number of events than a very large number. |
| **Binomial Distribution** | Its skewness depends on the probability of success `p` and the number of trials `n`. <br> *   If `p = 0.5`, the distribution is perfectly **symmetrical** (zero skew). <br> *   If `p < 0.5`, the distribution is **positively skewed**. <br> *   If `p > 0.5`, the distribution is **negatively skewed**. <br> As `n` increases, the Binomial distribution also tends to become more symmetrical, especially when `p` is not too far from 0.5, eventually approximating a Normal distribution. |
| **Exponential Distribution** | Always **positively skewed**. It models the time until an event occurs in a Poisson process, meaning there are many short durations and fewer long durations.                                                                                                                                                                          |
| **Chi-Squared Distribution** | Is also always **positively skewed**. The skewness decreases as the degrees of freedom increase, and it approaches a Normal distribution.                                                                                                                                                                                            |

### **Real-Life Examples of Skewed Distributions**

*   **Positive Skew (Right-Skewed):**
    *   **Household Income:** Most households earn a moderate income, but a small number of high-income earners pull the mean income higher than the median, creating a long tail to the right.
    *   **Waiting Times:** The time customers spend waiting in a queue; most people wait for a short period, but a few might experience very long waits.
    *   **Test Scores on a Difficult Exam:** Many students score low, but a few high achievers create a tail to the right.

*   **Negative Skew (Left-Skewed):**
    *   **Test Scores on an Easy Exam:** Most students score high, but a few struggling students pull the mean lower than the median, creating a tail to the left.
    *   **Lifespan of a Product:** Many products last a long time, but a small number fail early, resulting in a tail on the left.

### **Why Skewness Matters**

*   **Understanding Data Characteristics:** Skewness provides critical insights into the shape of the data distribution, revealing whether data points are concentrated towards higher or lower values.
*   **Impact on Central Tendency:** It highlights that the mean, median, and mode can differ significantly in skewed distributions, making the median a more robust measure of central tendency for skewed data, as it is less affected by extreme values.
*   **Assumptions for Statistical Tests:** Many statistical methods (e.g., t-tests, ANOVA) assume that data is normally distributed (i.e., symmetrical). Skewed data can violate these assumptions, leading to inaccurate results or requiring data transformation (e.g., logarithmic transformation) or the use of non-parametric tests.
*   **Machine Learning Model Performance:** Some machine learning algorithms perform better with normally distributed or symmetrical data. Addressing skewness through feature engineering (e.g., transformation) can improve model accuracy and stability.
*   **Risk Management:** In finance, understanding the skewness of returns is crucial. Negative skewness, for instance, implies a higher probability of large negative returns than large positive returns, which is a key consideration for investors.

By analyzing skewness, we gain a deeper understanding of the underlying data generation process and can make more informed decisions regarding data analysis, statistical modeling, and machine learning applications.

# **Python Code for Probability Distributions**

**Python Code using iris[1].csv**

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

# Load dataset
df = pd.read_csv('/content/iris[1].csv')

# Display first rows
print(df.head())
```

**1. Binomial Distribution**

```python
# Create binary data
mean_val = df['sepal_length'].mean()
df['success'] = df['sepal_length'] > mean_val

# Count successes
n = len(df)
k = df['success'].sum()
p = k / n

# Binomial PMF
x = np.arange(0, n+1)
y = binom.pmf(x, n, p)

# Plot (limited range for visibility)
plt.figure()
plt.bar(x[:20], y[:20])
plt.title("Binomial Distribution (Sepal Length > Mean)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()
```

**2. Poisson Distribution**

```python
# Count values per bin (simulate event counts)
counts, bins = np.histogram(df['sepal_length'], bins=10)

# Mean rate
lam = np.mean(counts)

# Poisson PMF
x = np.arange(0, max(counts)+1)
y = poisson.pmf(x, lam)

# Plot
plt.figure()
plt.bar(x, y)
plt.title("Poisson Distribution (Binned Counts)")
plt.xlabel("Event Count")
plt.ylabel("Probability")
plt.show()
```

**3. Normal Distribution**

```python
# Use sepal length (continuous data)
data = df['sepal_length']

# Parameters
mu = data.mean()
sigma = data.std()

# Generate values
x = np.linspace(data.min(), data.max(), 100)
y = norm.pdf(x, mu, sigma)

# Plot
plt.figure()
plt.hist(data, bins=20, density=True)
plt.plot(x, y)
plt.title("Normal Distribution - Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Density")
plt.show()
```

**4.  Comparison**

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

# Load dataset
df = pd.read_csv('/content/iris[1].csv')

# -------------------------------
# 1. NORMAL DISTRIBUTION (Direct)
# -------------------------------
data = df['sepal_length']
mu = data.mean()
sigma = data.std()

x_norm = np.linspace(data.min(), data.max(), 100)
y_norm = norm.pdf(x_norm, mu, sigma)

# -------------------------------
# 2. BINOMIAL DISTRIBUTION
# -------------------------------
mean_val = data.mean()
df['success'] = data > mean_val

n = len(df)
p = df['success'].sum() / n

x_bin = np.arange(0, 20)   # limited for visibility
y_bin = binom.pmf(x_bin, n, p)

# -------------------------------
# 3. POISSON DISTRIBUTION
# -------------------------------
counts, _ = np.histogram(data, bins=10)
lam = np.mean(counts)

x_pois = np.arange(0, max(counts)+1)
y_pois = poisson.pmf(x_pois, lam)

# -------------------------------
# PLOTTING COMPARISON
# -------------------------------
plt.figure()

# Normal
plt.plot(x_norm, y_norm, label='Normal Distribution')

# Binomial
plt.bar(x_bin, y_bin, alpha=0.5, label='Binomial')

# Poisson
plt.bar(x_pois, y_pois, alpha=0.5, label='Poisson')

plt.title("Comparison of Distributions on Iris Dataset")
plt.xlabel("Values / Counts")
plt.ylabel("Probability / Density")
plt.legend()
plt.show()
```
