#  **Q2. Why is NumPy considered faster than pure Python computations?**

NumPy is considered faster than pure Python computations because it is designed to perform numerical operations efficiently using optimized techniques and low-level implementations. While Python uses slower, high-level loops, NumPy performs operations in a faster and more optimized way.

---

##  **Main Reasons for High Performance**

### 1. Vectorization

NumPy uses **vectorized operations**, meaning it performs operations on entire arrays at once instead of using loops.

 This reduces:

* Execution time
* Code complexity

---

### 2. Implementation in C Language

NumPy is built using **C and C++**, which are much faster than Python.

 Advantages:

* Faster execution
* Efficient memory usage

---

### 3. Contiguous Memory Allocation

NumPy arrays store data in **continuous memory locations**, unlike Python lists.

 Benefits:

* Faster access
* Better cache performance

---

### 4. Reduced Overhead

Python lists store different data types and require type checking, while NumPy arrays:

* Store same data type
* Avoid repeated type checking

---

### 5. Built-in Optimized Functions

NumPy provides highly optimized mathematical functions that run faster than manual implementations.

---

##  **Table: NumPy vs Python**

| Feature        | Python Lists      | NumPy Arrays      |
| -------------- | ----------------- | ----------------- |
| Speed          | Slow (uses loops) | Fast (vectorized) |
| Memory Usage   | High              | Low               |
| Data Type      | Mixed             | Same type         |
| Implementation | Pure Python       | C-based           |
| Performance    | Low               | High              |

---

##  **Example: Performance Comparison**

```python
import numpy as np
import time

# Python list
lst = list(range(1000000))
start = time.time()
result = [x * 2 for x in lst]
print("Python time:", time.time() - start)

# NumPy array
arr = np.array(lst)
start = time.time()
result = arr * 2
print("NumPy time:", time.time() - start)
```

 Output shows NumPy is significantly faster.

---

##  **Real-Life Explanation**

Imagine calculating marks for 1 million students:

* Python: Processes one by one (slow)
* NumPy: Processes all at once (fast)

* Handles large datasets efficiently

* Ideal for real-time computations

#  **Q4. Explain Indexing and Slicing in NumPy Arrays**

Indexing and slicing are powerful techniques in NumPy used to access and manipulate specific elements or subsets of an array. These operations help in efficiently handling data without modifying the original structure unnecessarily.

---

##  **1. Indexing in NumPy**

Indexing is used to access a **single element** from an array using its position.

###  Types of Indexing:

### **a. 1D Array Indexing**

```python
import numpy as np

arr = np.array([10, 20, 30, 40])
print(arr[0])   # First element
print(arr[2])   # Third element
```

---

### **b. 2D Array Indexing**

```python
arr2 = np.array([[1, 2], [3, 4]])

print(arr2[0, 1])  # Row 0, Column 1 → 2
```

---

### **c. Negative Indexing**

Access elements from the end.

```python
print(arr[-1])  # Last element
```

---

##  **2. Slicing in NumPy**

Slicing is used to access **multiple elements (a subset)** of an array.

###  Syntax:

```
array[start : end : step]
```

---

###  Examples:

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])   # Elements from index 1 to 3
print(arr[:3])    # First 3 elements
print(arr[::2])   # Step of 2
```

---

###  2D Array Slicing:

```python
arr2 = np.array([[1,2,3],
                 [4,5,6]])

print(arr2[:, 1:3])  # All rows, columns 1 to 2
```

---

##  **3. Boolean Indexing**

Used to filter elements based on conditions.

```python
arr = np.array([10, 20, 30, 40])

print(arr[arr > 20])  # Elements greater than 20
```

---

##  **4. Fancy Indexing**

Uses a list of indices to access elements.

```python
arr = np.array([10, 20, 30, 40])

print(arr[[0, 2]])  # Access index 0 and 2
```

---

##  **Table: Indexing vs Slicing**

| Feature     | Indexing              | Slicing                  |
| ----------- | --------------------- | ------------------------ |
| Purpose     | Access single element | Access multiple elements |
| Syntax      | `arr[i]`              | `arr[start:end]`         |
| Output      | Scalar value          | Array                    |
| Flexibility | Less                  | More                     |

---

## 🔹 **Advantages of Indexing and Slicing**

* Easy data access
* Efficient data manipulation
* Saves time and memory
* Works with multi-dimensional arrays

---

## 🔹 **Real-Life Example**

If an array represents marks of students:

* Indexing → Get marks of one student
* Slicing → Get marks of a group of students

# **Bright Students (Advanced)**

**11. Implement vectorized operations for performance comparison**

````python
import numpy as np
import time

arr = np.arange(1000000)

# Without vectorization
start = time.time()
result = []
for i in arr:
    result.append(i * 2)
print("Loop time:", time.time() - start)

# With vectorization
start = time.time()
result = arr * 2
print("Vectorized time:", time.time() - start)
````
